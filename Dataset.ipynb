```python
import os
import shutil
from sklearn.model_selection import train_test_split

# =====================================================
# CONFIGURATION
# =====================================================
base_dir = r"Dataset"
train_src = os.path.join(base_dir, "Train")

# Output directories
train_dir = os.path.join(base_dir, "train")
valid_dir = os.path.join(base_dir, "valid")
test_dir  = os.path.join(base_dir, "test")

# Create directories if not exist
for d in [train_dir, valid_dir, test_dir]:
    os.makedirs(d, exist_ok=True)

# =====================================================
# SPLIT RATIOS
# =====================================================
train_ratio = 0.7
valid_ratio = 0.15
test_ratio  = 0.15

# =====================================================
# SPLIT DATA PER CLASS
# =====================================================
class_names = os.listdir(train_src)

for cls in class_names:
    src_path = os.path.join(train_src, cls)
    if not os.path.isdir(src_path):
        continue

    # Make subfolders
    os.makedirs(os.path.join(train_dir, cls), exist_ok=True)
    os.makedirs(os.path.join(valid_dir, cls), exist_ok=True)
    os.makedirs(os.path.join(test_dir, cls), exist_ok=True)

    images = [f for f in os.listdir(src_path) if os.path.isfile(os.path.join(src_path, f))]
    train_files, temp_files = train_test_split(images, test_size=(1 - train_ratio), random_state=42)
    valid_files, test_files = train_test_split(temp_files, test_size=(test_ratio / (test_ratio + valid_ratio)), random_state=42)

    # Copy files to each split
    for f in train_files:
        shutil.copy(os.path.join(src_path, f), os.path.join(train_dir, cls, f))
    for f in valid_files:
        shutil.copy(os.path.join(src_path, f), os.path.join(valid_dir, cls, f))
    for f in test_files:
        shutil.copy(os.path.join(src_path, f), os.path.join(test_dir, cls, f))

    print(f"Class '{cls}': {len(train_files)} train, {len(valid_files)} valid, {len(test_files)} test")

print(" Dataset split completed successfully.")
```
